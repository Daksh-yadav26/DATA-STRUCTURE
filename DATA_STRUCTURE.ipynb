2 Explain the difference between mutable and immutable data types with examples.

Mutable: Can be changed after creation.

Examples: list, dict, set

Example:

my_list = [1, 2, 3]
my_list[0] = 100  # Valid


Immutable: Cannot be changed after creation.

Examples: int, float, str, tuple

Example:

my_str = "hello"
# my_str[0] = "H"  # Invalid: strings are immutable

4 Describe how dictionaries store data.

Dictionaries store data as key-value pairs, using a hash table internally. Each key is hashed to a location in memory where its value is stored.

Keys must be unique and immutable.

Values can be of any type.

 5 Why might you use a set instead of a list in Python?

Use a set when:

You need unique elements (no duplicates).

You want faster membership testing (in operator is faster in sets).

6 What is a string in Python, and how is it different from a list?

A string is a sequence of characters, like "hello".

A list is a sequence of elements that can be of any data type.

Key differences:

Strings are immutable, lists are mutable.

Strings only hold characters, lists can hold anything.

7 How do tuples ensure data integrity in Python?

Since tuples are immutable, their contents cannot be changed after creation. This:

Prevents accidental modification.

Makes them useful for constants and hashable keys in dictionaries.

8. What is a hash table, and how does it relate to dictionaries in Python?

A hash table is a data structure that maps keys to values using a hash function.

Python's dict is implemented using a hash table.

It allows fast access, insertion, and deletion of items by key.

9 Can lists contain different data types in Python?

Yes, lists in Python are heterogeneous, meaning they can contain elements of different data types.

10 Why are strings immutable in Python?

Strings are immutable in Python for performance, security, and consistency reasons:

Once created, a string cannot be changed.

This allows strings to be hashable, which makes them usable as keys in dictionaries.

It also enables memory efficiency, as Python can safely reuse string object

11 What advantages do dictionaries offer over lists for certain tasks?

Dictionaries offer:

Fast lookups using keys (O(1) time complexity).

Meaningful associations (key-value pairs), which is clearer than using index-based access.

Dynamic and flexible data storage, such as storing user profiles or configurations.

14 How does the in keyword work differently for lists and dictionaries?

List: in checks if a value exists.

Dictionary: in checks if a key exists, not a value
